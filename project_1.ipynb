```python
df.drop(columns=["Status","unnamed1" ],inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
df.rename(columns={"Occupation": "Sector"},inplace=True)
```

```python
df["Marital_Status"] = df["Marital_Status"].astype(str).str.replace("0","Single")
```

```python
df["Marital_Status"] = df["Marital_Status"].astype(str).str.replace("1","Married")
```

# DATA VIAUALIZATION

```python
plt.figure(figsize=(15,6))

C1 =["#D7C57C", "#D1A92E"]

Chart_2 = sns.countplot(x="Age Group",data= df,hue="Marital_Status",palette=C1)

for i in Chart_2.containers:
    Chart_2.bar_label(i)


plt.title(" Age group wise Marital status", fontsize = 16,fontweight = "bold", family = "Georgia",color ="#80661A" )

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(10,5))

C2 = "#2F4B6E", "#6B9AD3"

Chart_1 = sns.countplot(x="Age Group",data=df,hue="Gender",palette=C2)

for i in Chart_1.containers:
    Chart_1.bar_label(i)

plt.title("Gender Wise Age Groups count",fontsize = 16,fontweight= "bold",family= "Georgia" , color="#547EB2")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
D1 = df.groupby("Zone",as_index=False)["Orders"].sum()

C3 = [ "#8A2D3C", "#D86572", "#FF9DA7", "#FFC9CF", "#FFEFF1"]

plt.pie(D1["Orders"],labels=D1["Zone"],autopct="%.f%%",colors=C3)

plt.title("Zone Wise Orders",fontsize = 16,fontweight= "bold",family= "Georgia" ,color = "#BA656F" )

plt.show()
```

```python
plt.figure(figsize=(15,5))

sns.lineplot(x="Sector",y="Amount",data=df,ci=None,marker="o",color ="#4F9450")

plt.title("Sector Wise Total Amount",fontsize = 16,fontweight= "bold",family= "Georgia" ,color ="#1A6C1C")

plt.xticks(rotation=45)

plt.grid()

plt.show()
```

```python
plt.figure(figsize =  (15,8))


sns.lineplot(x="State",y="Orders",data=df,ci=None,marker="o",color = "#855743")

plt.xticks(rotation = 45)

plt.title("State Wise Orders",fontsize = 16,fontweight= "bold",family= "Georgia" ,color ="#442315")

plt.xlabel("")

plt.ylabel("")

plt.grid()

plt.show()
```
